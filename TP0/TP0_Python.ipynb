####**Tarea**: generar un vector con n elementos generados al azar.

```
import random

def aleatorio_1(n):
    """Método 1: dinámico usando bucle for
    """
    v = []
    for i in range(n):
      v.append(random.random())
    return v

def aleatorio_2(n):
    """Método 2:  inicializando e iterando
    """
    v = [0]*n
    for i in range(len(v)):
      v[i] = random.random()
    return v

def aleatorio_3(n):
    """Método 2: list comprehensions
    """
    return [random.random() for i in range(n)]

n = 6
print('Método 1 (con for dinámico)')
print(aleatorio_1(n))
print('Método 2 (con inicializacion)')
print(aleatorio_2(n))
print('Método 3 (list comprehensions)')
print(aleatorio_3(n))
```

## Ejercicio 2
Resolver los siguientes ejercicios de programación.

1.    Implementar un programa que iterativamente genere numeros aleatorios enteros dentro del intervalo [0, 20] y que termine la ejecución al obtener un 0. Debe imprimir en pantalla la cantidad de iteraciones realizadas al terminar la ejecución.

2.    Generar un vector con n elementos binarios (valor 0 o 1 generado al azar). Luego contar cuantos ceros hay en la primera mitad del vector (posición 0 a n/2) y cuantos unos hay en la segunda mitad del vector. Probar con n = 10 y n = 13.


**Ejercicio 1:** Generación de números aleatorios enteros dentro del intervalo **[0, 20]**, con finalización de ejecución al obtener un 0

```
from random import randint

i = 0

while True:
    x = randint(0, 20) # Generación de número aleatorio dentro de [0, 20]
    i += 1 # Número de iteración
    print("Número aleatorio: {0} ; Iteración: {1}".format(x, i))
    if x == 0:
      break # Sale del bucle while si el número generado es 0

print("\n******************************")
print("Cantidad de iteraciones:", i)
print("******************************")
```

**Ejercicio 2:** Generación de un vector con n elementos binarios, y conteo de 0s y 1s en primera y segunda mitad del vector, respectivamente

```
from random import random
from math import ceil

n, cant_cero, cant_uno = 10, 0, 0
mitad = ceil(n/2)
vector = []

# Generación del vector con n elementos
for _ in range(n):
    vector.append(round(random()))

# Recorrido por el vector: se usa enumerate() para obtener el índice y valor
# de cada ítem
for i, v in enumerate(vector):
    if i < mitad: # Verificación en 1era mitad
        if v == 0:
            cant_cero += 1
    else: # Verificación en 2da mitad
        if v == 1:
            cant_uno += 1

print("Vector:", vector)
print("Cant. ceros en 1era mitad:", cant_cero)
print("Cant. unos en 2da mitad:", cant_uno)

```
