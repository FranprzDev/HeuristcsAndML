## Ejercicio 1
Dada la ecuación de la recta, se solicita emplear la librería Matplotlib (o alguna de su elección) para graficar a través de una función los puntos de la misma.

####**Tarea**: Crear una recta de 20 elementos.

```
print('Configurar parámetros de la recta.')
cant_elementos = 20
a = 2
b = 10
recta = [a*x+b for x in range(cant_elementos)]

print(f"Valores de la recta: {recta}")
```

####**Tarea**: Usar la librería de Matplotlib para graficar la recta.

```
import matplotlib.pyplot as plt
%matplotlib inline

x_values_recta = [i for i in range(cant_elementos)]
plt.plot(x_values_recta,recta,'ro-',label=f'recta simple a={a} - b={b}')
plt.xlabel('Valores de X')
plt.ylabel('Valores de Y')
plt.title(f"Recta en el plano: pendiente {a} y ordenada al origin {b}.")
```

## Ejercicio 2
Graficar la norma del vector para una corrida con 10, 50, 100, y 500 puntos aleatorios. De ser necesario, buscar lo que se denomina "norma de un vector".

```
import random
import math
import matplotlib.pyplot as plt

def random_gen(n):
    return [(random.random(), random.random()) for i in range(n)]

def calculate_norm(vector):
    return math.sqrt(vector[0]**2 + vector[1]**2)

quantitys = [10,50,100,500]

vector_10 = random_gen(quantitys[0])
vector_50 = random_gen(quantitys[1])
vector_100 = random_gen(quantitys[2])
vector_500 = random_gen(quantitys[3])

vectors = [vector_10, vector_50, vector_100, vector_500]
norms = []

for v in vectors:
  norms_of_v = []
  for point in v:
    norm = calculate_norm(point)
    norms_of_v.append(norm)
  norms.append(norms_of_v)

# zip agrupa; como n = 10 y normas =  [10 elementos] -> [(1, norma_1),(2, norma_2)]
for n, normas in zip(quantitys, norms):
    plt.figure()
    plt.plot(range(1, n+1), normas, "go")
    plt.title(f"Normas de {n} puntos aleatorios")
    plt.xlabel("Número de punto")
    plt.ylabel("Norma")
    plt.grid()
    plt.show()
```
