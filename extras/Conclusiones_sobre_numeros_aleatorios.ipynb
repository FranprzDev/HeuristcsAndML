El siguiente código es para demostrar que no es adecuado usar "round" para generar numeros enteros aleatorios.

```
import random
import math
import matplotlib.pyplot as plt

#CONSIGNA:
#generar N numeros ENTEROS aleatorios con probabilidad uniforme, entre a y b

# round -> Redondea al valor más proximo
# int -> Lo convierte en un entero (trunca)
# ceil -> Redondea para arriba
# Randint -> Genera un número aleatorio entero (UNIFORME YA HECHA)

def generar1(N, a, b):
  # Intervalo de generación [2;6]
  return [round(random.random()*(b-a)+a) for i in range(N)]

def generar2(N, a, b):
  # Intervalo de generación [2;7)
  return [int(random.random()*(1+b-a)+a) for i in range(N)]

def generar3(N, a, b):
  # Intervalo de generación [1,6)
  return [math.ceil(random.random()*(1+b-a)+a-1) for i in range(N)]

def generar4(N, a, b):
  # Intervalo de generación [2,6]
  return [random.randint(a, b) for i in range(N)]

#contar las veces que sale cada numero (cuantos numeros 0, cuantos 1, etc.)
def contar(vector, texto):
  print(vector)
  n = max(vector)+1
  cantidad = [0]*n
  for i in vector:
      cantidad[i] = cantidad[i] + 1

  plt.bar(range(n), cantidad)
  plt.title(texto)
  plt.xlabel("Número")
  plt.ylabel("Cantidad de veces")
  plt.show()
```

Al generar con método con "round" la probabilidad no es uniforme (los extremos tienen una probabilidad menor). Con "int", en cambio, sí se logra una probabilidad uniforme.

```
N = 10
vector = generar1(N, 2, 6)
# 2.5 -> 2
# 2.51 -> 3
# [2.51, 3,49] -> 3
# Usando round, la diferencia está en los extremos, cuando utilizamos un extremos tenemos menos probabilidad
# de que toque debido a ej: 2; [2;2.5] en cambio si utilizamos alguno interno; ej 3; su rango estará en [2.51;3.49];
# por lo tanto es más grande el intervalo del valor vecino al extremo (casi el doble??)
# Intervalo de generación [2;6]
contar(vector, 'Cantidad de veces que sale cada numero, usando "round" (ERROR)')

vector = generar2(N, 2, 6)
# Este trunca, 2 -> [2;3) -> Todos tendrán el mismo rango de probabilidad.
# Intervalo de generación [2;7)
contar(vector, 'Cantidad de veces que sale cada numero, usando "int" (BIEN)')

vector = generar3(N, 2, 6)
# Todos tienen la misma probabilidad. equiprobable.
# Todos tienen la misma probaiblidad, debido a que se utiliza el siguiente intervalo [1,6)
contar(vector, 'Cantidad de veces que sale cada numero, usando "ceil" (BIEN)')

vector = generar4(N, 2, 6)
# Es una función de librería uniforme de toda la vida.
contar(vector, 'Cantidad de veces que sale cada numero, usando "randint" (BIEN)')
```
